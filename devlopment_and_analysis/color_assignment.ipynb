```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
```

```python
input_path="../output_video/cropped_image.jpg"
image=cv2.imread(input_path)
image=cv2.cvtColor(image,cv2.COLOR_BGR2RGB)
```

```python
plt.imshow(image)
plt.show()
```

```python
half_image=image[0:image.shape[0]//2,:,:]
plt.imshow(half_image)
plt.show()
```

```python
image_2d=half_image.reshape(-1,3)
kmeans=KMeans(n_clusters=2, random_state=0).fit(image_2d)
```

```python
labels=kmeans.labels_
clustered_img=labels.reshape(half_image.shape[0],half_image.shape[1])
plt.imshow(clustered_img)
plt.show()
```

```python
corner_cluster=[clustered_img[0,0],clustered_img[0,-1],clustered_img[-1,0],clustered_img[-1,-1]]
non_player_cluster=max(set(corner_cluster), key=corner_cluster.count)
print(non_player_cluster)
```

```python
player_cluster=1-non_player_cluster
kmeans.cluster_centers_[player_cluster]
```
